```python

import librosa
import librosa.display

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

from pathlib import Path
```

```python
PROJECT_ROOT = Path("..")

METADATA_PATH = PROJECT_ROOT / "data" / "metadata" / "metadata.csv"

metadata = pd.read_csv(METADATA_PATH)

metadata.head()
```

```python
sample = metadata.iloc[0]

sample
```

```python
audio_path = Path(sample["filepath"])

audio_path
```

```python
audio, sr = librosa.load(audio_path, sr=None)
```

```python
audio
```

```python
sr
```

```python
print(type(audio))

print(type(sr))
```

```python
print(audio.shape)
```

```python
duration = len(audio) / sr

print(duration)
```

```python
audio[:20]
```

```python
plt.figure(figsize=(14,4))

librosa.display.waveshow(audio, sr=sr)

plt.title("Waveform")

plt.xlabel("Time (seconds)")

plt.ylabel("Amplitude")

plt.show()
```

```python
print(audio.max())
```

```python
print(audio.min())
```

```python
plt.figure(figsize=(16,4))

plt.plot(audio[:3000])

plt.title("First 3000 Samples")

plt.show()
```
